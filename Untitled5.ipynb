```python
# ============================================================
# Simulador de Sincronización Discreta tipo Kuramoto sobre Grafos
# Versión corregida conforme al paper
# ============================================================

import numpy as np
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display, clear_output
from itertools import product


# ============================================================
# 1. Funciones auxiliares
# ============================================================

def bell_number(n):
    """
    Calcula el número de Bell B_n.
    B_n cuenta el número de particiones de un conjunto con n elementos.
    """
    bell = [[0 for _ in range(n + 1)] for _ in range(n + 1)]
    bell[0][0] = 1

    for i in range(1, n + 1):
        bell[i][0] = bell[i - 1][i - 1]
        for j in range(1, i + 1):
            bell[i][j] = bell[i - 1][j - 1] + bell[i][j - 1]

    return bell[n][0]


def kuramoto_order_parameter(theta, M):
    """
    Parámetro de orden tipo Kuramoto para fases discretas.

    r(t) = |(1/N) sum_j exp(2 pi i theta_j / M)|

    r = 1 indica sincronización perfecta.
    r cercano a 0 indica dispersión de fases.
    """
    theta = np.array(theta, dtype=int)
    z = np.exp(2j * np.pi * theta / M)
    return np.abs(np.mean(z))


# ============================================================
# 2. Clase principal del modelo
# ============================================================

class KuramotoDiscreto:
    """
    Red discreta de Kuramoto con umbral sobre un grafo finito.

    Modelo:

        (F_kappa(theta))_i = theta_i + omega* + sigma_kappa(S_i(theta)) mod M

    donde:

        S_i(theta) = sum_{j in N(i)} d_M(theta_j, theta_i)

    y:

        d_M(a,b) = ((a-b+floor(M/2)) mod M) - floor(M/2)

    Convención corregida conforme al paper:

        sigma_kappa(S) = +1 si S >= kappa
                       = -1 si S <= -kappa
                       =  0 si |S| < kappa
    """

    def __init__(self, N=4, M=6, kappa=2, omega=0, tipo_grafo="Completo", epsilon=1):
        self.N = int(N)
        self.M = int(M)
        self.kappa = int(kappa)
        self.omega = int(omega)
        self.tipo_grafo = tipo_grafo
        self.epsilon = float(epsilon)
        self.G = self.crear_grafo()
        self.pos = nx.spring_layout(self.G, seed=7)

    # --------------------------------------------------------
    # Grafo
    # --------------------------------------------------------
    def crear_grafo(self):
        if self.tipo_grafo == "Completo":
            return nx.complete_graph(self.N)
        elif self.tipo_grafo == "Camino":
            return nx.path_graph(self.N)
        elif self.tipo_grafo == "Ciclo":
            return nx.cycle_graph(self.N)
        elif self.tipo_grafo == "Estrella":
            return nx.star_graph(self.N - 1)
        else:
            raise ValueError("Tipo de grafo no reconocido. Usa: Completo, Camino, Ciclo o Estrella.")

    # --------------------------------------------------------
    # Distancia angular con signo
    # --------------------------------------------------------
    def dM_signed(self, a, b):
        """
        Diferencia angular con signo:

            d_M(a,b) = ((a-b+floor(M/2)) mod M) - floor(M/2)

        Interpreta el desplazamiento orientado desde b hacia a.
        """
        return ((int(a) - int(b) + self.M // 2) % self.M) - self.M // 2

    def dM_circular(self, a, b):
        """
        Distancia circular sin signo:

            d_M(a,b) = min(|a-b|, M-|a-b|)

        Esta distancia se usa para construir la subred epsilon-sincronizada.
        """
        diff = abs(int(a) - int(b))
        return min(diff, self.M - diff)

    # --------------------------------------------------------
    # Función signo umbralizada
    # --------------------------------------------------------
    def sigma_kappa(self, S):
        """
        Función signo umbralizada con convención incluyente.

        sigma_kappa(S) = +1 si S >= kappa
                       = -1 si S <= -kappa
                       =  0 si |S| < kappa
        """
        if S >= self.kappa:
            return 1
        elif S <= -self.kappa:
            return -1
        else:
            return 0

    # --------------------------------------------------------
    # Suma de acoplamiento
    # --------------------------------------------------------
    def S_i(self, theta, i):
        """
        Suma de acoplamiento con signo del vértice i:

            S_i(theta) = sum_{j in N(i)} d_M(theta_j, theta_i)
        """
        theta = np.array(theta, dtype=int)
        return sum(self.dM_signed(theta[j], theta[i]) for j in self.G.neighbors(i))

    def vector_S(self, theta):
        """
        Vector de sumas de acoplamiento:

            S(theta) = (S_1(theta), ..., S_N(theta))
        """
        return np.array([self.S_i(theta, i) for i in range(self.N)], dtype=int)

    def vector_sigma(self, theta):
        """
        Vector de correcciones sigma_kappa(S_i(theta)).
        """
        S = self.vector_S(theta)
        return np.array([self.sigma_kappa(s) for s in S], dtype=int)

    # --------------------------------------------------------
    # Dinámica F_kappa
    # --------------------------------------------------------
    def paso(self, theta):
        """
        Aplica una iteración de F_kappa a la configuración theta.
        """
        theta = np.array(theta, dtype=int)
        sigma = self.vector_sigma(theta)
        nueva = (theta + self.omega + sigma) % self.M
        return nueva.astype(int)

    def simular(self, theta0, T=30):
        """
        Simula la trayectoria theta^0, theta^1, ..., theta^T.
        """
        theta = np.array(theta0, dtype=int) % self.M

        trayectoria = [theta.copy()]
        S_hist = []
        sigma_hist = []
        r_hist = [kuramoto_order_parameter(theta, self.M)]
        subredes = [self.subred_acuerdo_edges(theta)]

        for _ in range(T):
            S_t = self.vector_S(theta)
            sigma_t = self.vector_sigma(theta)

            S_hist.append(S_t.copy())
            sigma_hist.append(sigma_t.copy())

            theta = self.paso(theta)

            trayectoria.append(theta.copy())
            r_hist.append(kuramoto_order_parameter(theta, self.M))
            subredes.append(self.subred_acuerdo_edges(theta))

        return {
            "trayectoria": np.array(trayectoria),
            "S_hist": np.array(S_hist),
            "sigma_hist": np.array(sigma_hist),
            "r_hist": np.array(r_hist),
            "subredes": subredes,
        }

    # --------------------------------------------------------
    # Sincronización y subred epsilon-sincronizada
    # --------------------------------------------------------
    def esta_sincronizado(self, theta):
        """
        Verifica si theta pertenece a la diagonal Delta_M.
        """
        theta = np.array(theta, dtype=int)
        return np.all(theta == theta[0])

    def subred_acuerdo_edges(self, theta):
        """
        Aristas de la subred epsilon-sincronizada:

            E_theta^epsilon = {(i,j) in E : d_M(theta_i, theta_j) < epsilon}

        Con epsilon = 1 se obtiene acuerdo exacto, porque la distancia
        circular toma valores enteros.
        """
        theta = np.array(theta, dtype=int)
        edges = []

        for i, j in self.G.edges():
            if self.dM_circular(theta[i], theta[j]) < self.epsilon:
                edges.append((i, j))

        return tuple(sorted(edges))

    # --------------------------------------------------------
    # Clasificación de atractores
    # --------------------------------------------------------
    def clasificar_atractor(self, theta0, T_max=200):
        """
        Clasifica la trayectoria como:

        - sync
        - punto fijo no trivial
        - ciclo
        - transitorio
        """
        theta = np.array(theta0, dtype=int) % self.M
        vistos = {}

        for t in range(T_max + 1):
            key = tuple(theta.tolist())

            if self.esta_sincronizado(theta):
                return {
                    "tipo": "sync",
                    "tiempo": t,
                    "periodo": 1,
                    "estado_final": theta.copy()
                }

            if key in vistos:
                t_inicio = vistos[key]
                periodo = t - t_inicio

                if periodo == 1:
                    return {
                        "tipo": "punto fijo no trivial",
                        "tiempo": t_inicio,
                        "periodo": 1,
                        "estado_final": theta.copy()
                    }
                else:
                    return {
                        "tipo": "ciclo",
                        "tiempo": t_inicio,
                        "periodo": periodo,
                        "estado_final": theta.copy()
                    }

            vistos[key] = t
            theta = self.paso(theta)

        return {
            "tipo": "transitorio",
            "tiempo": T_max,
            "periodo": None,
            "estado_final": theta.copy()
        }

    # --------------------------------------------------------
    # Tabla de trayectoria
    # --------------------------------------------------------
    def tabla_trayectoria(self, resultado):
        """
        Construye una tabla con:

        - t
        - theta(t)
        - r(t)
        - S_i(theta_t)
        - sigma_kappa(S_i)
        - subred de acuerdo
        """
        trayectoria = resultado["trayectoria"]
        S_hist = resultado["S_hist"]
        sigma_hist = resultado["sigma_hist"]
        r_hist = resultado["r_hist"]
        subredes = resultado["subredes"]

        filas = []

        for t in range(len(trayectoria)):
            fila = {
                "t": t,
                "theta(t)": tuple(trayectoria[t]),
                "r(t)": round(float(r_hist[t]), 4),
                "subred_acuerdo": subredes[t]
            }

            if t < len(S_hist):
                fila["S_i(theta_t)"] = tuple(S_hist[t])
                fila["sigma_kappa(S_i)"] = tuple(sigma_hist[t])
            else:
                fila["S_i(theta_t)"] = "-"
                fila["sigma_kappa(S_i)"] = "-"

            filas.append(fila)

        return pd.DataFrame(filas)


# ============================================================
# 3. Exploración exhaustiva del espacio X_M = Z_M^V
# ============================================================

def exploracion_exhaustiva(N, M, kappa, omega=0, tipo_grafo="Completo", epsilon=1, T_max=150):
    """
    Recorre todo el espacio de configuraciones M^N y calcula:

    - rho_sync
    - rho_fix
    - rho_cyc
    - rho_transitorio
    - tiempos de sincronización
    - subredes realizables
    - transiciones entre subredes
    - comparación con número de Bell B_N
    """
    modelo = KuramotoDiscreto(
        N=N,
        M=M,
        kappa=kappa,
        omega=omega,
        tipo_grafo=tipo_grafo,
        epsilon=epsilon
    )

    total = M ** N

    conteo = {
        "sync": 0,
        "punto fijo no trivial": 0,
        "ciclo": 0,
        "transitorio": 0
    }

    tiempos_sync = []
    subredes_realizables = set()
    transiciones = set()

    for theta0 in product(range(M), repeat=N):
        clasif = modelo.clasificar_atractor(theta0, T_max=T_max)
        conteo[clasif["tipo"]] += 1

        if clasif["tipo"] == "sync":
            tiempos_sync.append(clasif["tiempo"])

        resultado = modelo.simular(theta0, T=min(T_max, 50))
        subredes = resultado["subredes"]

        for s in subredes:
            subredes_realizables.add(s)

        for a, b in zip(subredes[:-1], subredes[1:]):
            if a != b:
                transiciones.add((a, b))

    resumen = {
        "total_configuraciones": total,
        "sync": conteo["sync"],
        "fix": conteo["punto fijo no trivial"],
        "cyc": conteo["ciclo"],
        "transitorio": conteo["transitorio"],
        "rho_sync": conteo["sync"] / total,
        "rho_fix": conteo["punto fijo no trivial"] / total,
        "rho_cyc": conteo["ciclo"] / total,
        "rho_transitorio": conteo["transitorio"] / total,
        "tiempos_sync": tiempos_sync,
        "num_subredes_realizables": len(subredes_realizables),
        "num_transiciones": len(transiciones),
        "Bell_BN": bell_number(N),
        "subredes_realizables": subredes_realizables,
        "transiciones": transiciones,
    }

    return resumen


# ============================================================
# 4. Barrido de kappa para mapa de regímenes
# ============================================================

def barrido_kappa(N, M, omega=0, tipo_grafo="Completo", epsilon=1, T_max=150):
    """
    Barre kappa desde 1 hasta degmax*floor(M/2).
    Calcula rho_sync, rho_fix y rho_cyc para cada kappa.
    """
    modelo_temp = KuramotoDiscreto(
        N=N,
        M=M,
        kappa=1,
        omega=omega,
        tipo_grafo=tipo_grafo,
        epsilon=epsilon
    )

    degmax = max(dict(modelo_temp.G.degree()).values())
    kappa_max = degmax * (M // 2)

    filas = []

    for kappa in range(1, kappa_max + 1):
        resumen = exploracion_exhaustiva(
            N=N,
            M=M,
            kappa=kappa,
            omega=omega,
            tipo_grafo=tipo_grafo,
            epsilon=epsilon,
            T_max=T_max
        )

        filas.append({
            "kappa": kappa,
            "rho_sync": resumen["rho_sync"],
            "rho_fix": resumen["rho_fix"],
            "rho_cyc": resumen["rho_cyc"],
            "rho_transitorio": resumen["rho_transitorio"],
            "subredes_realizables": resumen["num_subredes_realizables"],
            "transiciones": resumen["num_transiciones"],
            "Bell_BN": resumen["Bell_BN"]
        })

    return pd.DataFrame(filas)


# ============================================================
# 5. Visualizaciones
# ============================================================

def plot_heatmap(resultado):
    trayectoria = resultado["trayectoria"]

    fig, ax = plt.subplots(figsize=(8, 4))
    im = ax.imshow(trayectoria.T, aspect="auto", interpolation="nearest")
    ax.set_xlabel("Tiempo t")
    ax.set_ylabel("Vértice i")
    ax.set_title("Heatmap de evolución de fases")
    cbar = plt.colorbar(im, ax=ax)
    cbar.set_label("Fase en Z_M")
    plt.show()


def plot_order_parameter(resultado):
    r = resultado["r_hist"]

    fig, ax = plt.subplots(figsize=(8, 3))
    ax.plot(range(len(r)), r, marker="o")
    ax.set_ylim(-0.05, 1.05)
    ax.set_xlabel("Tiempo t")
    ax.set_ylabel("r(t)")
    ax.set_title("Parámetro de orden de Kuramoto")
    ax.grid(True, alpha=0.3)
    plt.show()


def plot_grafo(modelo, theta, title="Grafo coloreado por fase"):
    theta = np.array(theta, dtype=int)

    fig, ax = plt.subplots(figsize=(5, 5))

    node_colors = theta

    nx.draw_networkx_edges(modelo.G, modelo.pos, ax=ax, alpha=0.25)

    acuerdo_edges = modelo.subred_acuerdo_edges(theta)
    nx.draw_networkx_edges(
        modelo.G,
        modelo.pos,
        edgelist=acuerdo_edges,
        ax=ax,
        width=3
    )

    nodes = nx.draw_networkx_nodes(
        modelo.G,
        modelo.pos,
        node_color=node_colors,
        cmap=plt.cm.viridis,
        vmin=0,
        vmax=modelo.M - 1,
        node_size=650,
        ax=ax
    )

    nx.draw_networkx_labels(modelo.G, modelo.pos, ax=ax)

    cbar = plt.colorbar(nodes, ax=ax)
    cbar.set_label("Fase")

    ax.set_title(title)
    ax.axis("off")
    plt.show()


def plot_hist_tiempos(tiempos):
    if len(tiempos) == 0:
        print("No hubo trayectorias sincronizadas; no hay histograma de tiempos.")
        return

    fig, ax = plt.subplots(figsize=(7, 3))
    ax.hist(
        tiempos,
        bins=range(min(tiempos), max(tiempos) + 2),
        align="left",
        rwidth=0.85
    )
    ax.set_xlabel("Tiempo de sincronización")
    ax.set_ylabel("Frecuencia")
    ax.set_title("Histograma de tiempos de sincronización")
    plt.show()


def plot_mapa_regimenes(df):
    fig, ax = plt.subplots(figsize=(8, 4))

    ax.plot(df["kappa"], df["rho_sync"], marker="o", label="rho_sync")
    ax.plot(df["kappa"], df["rho_fix"], marker="o", label="rho_fix")
    ax.plot(df["kappa"], df["rho_cyc"], marker="o", label="rho_cyc")

    ax.set_xlabel("kappa")
    ax.set_ylabel("Proporción")
    ax.set_title("Mapa de regímenes por barrido de kappa")
    ax.set_ylim(-0.05, 1.05)
    ax.grid(True, alpha=0.3)
    ax.legend()

    plt.show()


def construir_diagrama_transiciones(transiciones):
    """
    Construye un grafo dirigido de transiciones entre subredes.

    Cada nodo es una subred, codificada como tupla de aristas.
    """
    D = nx.DiGraph()

    for a, b in transiciones:
        D.add_node(a)
        D.add_node(b)
        D.add_edge(a, b)

    return D


def plot_diagrama_transiciones(transiciones, max_labels=20):
    """
    Grafica el diagrama de transiciones entre subredes.
    Recomendado sólo para casos pequeños.
    """
    D = construir_diagrama_transiciones(transiciones)

    if D.number_of_nodes() == 0:
        print("No hay transiciones no triviales entre subredes.")
        return

    fig, ax = plt.subplots(figsize=(8, 6))
    pos = nx.spring_layout(D, seed=11)

    nx.draw_networkx_nodes(D, pos, node_size=500, ax=ax)
    nx.draw_networkx_edges(D, pos, arrows=True, arrowstyle="->", arrowsize=15, ax=ax)

    if D.number_of_nodes() <= max_labels:
        labels = {node: str(node) for node in D.nodes()}
        nx.draw_networkx_labels(D, pos, labels=labels, font_size=7, ax=ax)

    ax.set_title("Diagrama de transiciones de subredes")
    ax.axis("off")
    plt.show()


# ============================================================
# 6. Interfaz con widgets
# ============================================================

N_widget = widgets.IntSlider(
    value=3,
    min=2,
    max=7,
    step=1,
    description="N"
)

M_widget = widgets.IntSlider(
    value=4,
    min=2,
    max=12,
    step=1,
    description="M"
)

kappa_widget = widgets.IntSlider(
    value=2,
    min=1,
    max=20,
    step=1,
    description="kappa"
)

omega_widget = widgets.IntSlider(
    value=0,
    min=0,
    max=11,
    step=1,
    description="omega*"
)

epsilon_widget = widgets.FloatSlider(
    value=1.0,
    min=0.1,
    max=6.0,
    step=0.1,
    description="epsilon"
)

T_widget = widgets.IntSlider(
    value=20,
    min=1,
    max=100,
    step=1,
    description="T"
)

grafo_widget = widgets.Dropdown(
    options=["Completo", "Camino", "Ciclo", "Estrella"],
    value="Completo",
    description="Grafo"
)

modo_theta_widget = widgets.Dropdown(
    options=["Ejemplo paper", "Aleatoria", "Manual simple"],
    value="Ejemplo paper",
    description="theta0"
)

boton_simular = widgets.Button(
    description="Simular trayectoria",
    button_style="success"
)

boton_exhaustivo = widgets.Button(
    description="Exploración exhaustiva",
    button_style="warning"
)

boton_barrido = widgets.Button(
    description="Barrido de kappa",
    button_style="info"
)

salida = widgets.Output()


# ------------------------------------------------------------
# Actualización de rangos
# ------------------------------------------------------------

def actualizar_rangos(*args):
    N = N_widget.value
    M = M_widget.value
    tipo = grafo_widget.value

    omega_widget.max = M - 1
    epsilon_widget.max = M // 2 + 1

    modelo_temp = KuramotoDiscreto(
        N=N,
        M=M,
        kappa=1,
        omega=0,
        tipo_grafo=tipo,
        epsilon=1
    )

    degmax = max(dict(modelo_temp.G.degree()).values())
    kappa_max = max(1, degmax * (M // 2) + 1)

    kappa_widget.max = kappa_max

    if kappa_widget.value > kappa_widget.max:
        kappa_widget.value = kappa_widget.max


for w in [N_widget, M_widget, grafo_widget]:
    w.observe(actualizar_rangos, names="value")

actualizar_rangos()


# ------------------------------------------------------------
# Condición inicial
# ------------------------------------------------------------

def construir_theta0(N, M, modo):
    if modo == "Ejemplo paper":
        # Ejemplo del paper: N=3, M=4, theta0=(0,2,3)
        base = [0, M // 2, M - 1]
        theta = (base + [0] * N)[:N]
        return np.array(theta, dtype=int) % M

    elif modo == "Manual simple":
        # Patrón simple determinista: 0,1,2,... módulo M
        return np.array([i % M for i in range(N)], dtype=int)

    else:
        return np.random.randint(0, M, size=N)


# ------------------------------------------------------------
# Acción: simulación individual
# ------------------------------------------------------------

def on_simular_clicked(b):
    with salida:
        clear_output(wait=True)

        N = N_widget.value
        M = M_widget.value
        kappa = kappa_widget.value
        omega = omega_widget.value
        epsilon = epsilon_widget.value
        T = T_widget.value
        tipo_grafo = grafo_widget.value
        modo = modo_theta_widget.value

        modelo = KuramotoDiscreto(
            N=N,
            M=M,
            kappa=kappa,
            omega=omega,
            tipo_grafo=tipo_grafo,
            epsilon=epsilon
        )

        theta0 = construir_theta0(N, M, modo)

        resultado = modelo.simular(theta0, T=T)
        clasif = modelo.clasificar_atractor(theta0, T_max=max(200, T))
        tabla = modelo.tabla_trayectoria(resultado)

        print("SIMULACIÓN INDIVIDUAL")
        print("=====================")
        print(f"Grafo: {tipo_grafo}")
        print(f"N = {N}")
        print(f"M = {M}")
        print(f"kappa = {kappa}")
        print(f"omega* = {omega}")
        print(f"epsilon = {epsilon}")
        print(f"theta0 = {tuple(theta0)}")
        print()
        print(f"Clasificación del atractor: {clasif['tipo']}")
        print(f"Tiempo detectado: {clasif['tiempo']}")
        print(f"Periodo: {clasif['periodo']}")
        print(f"Estado final detectado: {tuple(clasif['estado_final'])}")
        print()

        plot_heatmap(resultado)
        plot_order_parameter(resultado)

        plot_grafo(modelo, resultado["trayectoria"][0], title="Snapshot inicial")

        mitad = len(resultado["trayectoria"]) // 2
        plot_grafo(modelo, resultado["trayectoria"][mitad], title=f"Snapshot t={mitad}")

        plot_grafo(modelo, resultado["trayectoria"][-1], title=f"Snapshot final t={T}")

        display(tabla)


# ------------------------------------------------------------
# Acción: exploración exhaustiva
# ------------------------------------------------------------

def on_exhaustivo_clicked(b):
    with salida:
        clear_output(wait=True)

        N = N_widget.value
        M = M_widget.value
        kappa = kappa_widget.value
        omega = omega_widget.value
        epsilon = epsilon_widget.value
        tipo_grafo = grafo_widget.value

        total = M ** N

        print("EXPLORACIÓN EXHAUSTIVA")
        print("======================")
        print(f"Total de configuraciones: M^N = {M}^{N} = {total}")
        print()

        if total > 20000:
            print("Advertencia: el espacio de configuraciones es grande.")
            print("Reduce N o M para una exploración más rápida.")
            return

        resumen = exploracion_exhaustiva(
            N=N,
            M=M,
            kappa=kappa,
            omega=omega,
            tipo_grafo=tipo_grafo,
            epsilon=epsilon,
            T_max=150
        )

        df_resumen = pd.DataFrame([{
            "total": resumen["total_configuraciones"],
            "sync": resumen["sync"],
            "fix": resumen["fix"],
            "cyc": resumen["cyc"],
            "transitorio": resumen["transitorio"],
            "rho_sync": resumen["rho_sync"],
            "rho_fix": resumen["rho_fix"],
            "rho_cyc": resumen["rho_cyc"],
            "rho_transitorio": resumen["rho_transitorio"],
            "subredes_realizables": resumen["num_subredes_realizables"],
            "transiciones": resumen["num_transiciones"],
            "Bell_BN": resumen["Bell_BN"],
        }])

        display(df_resumen)
        plot_hist_tiempos(resumen["tiempos_sync"])
        plot_diagrama_transiciones(resumen["transiciones"])


# ------------------------------------------------------------
# Acción: barrido de kappa
# ------------------------------------------------------------

def on_barrido_clicked(b):
    with salida:
        clear_output(wait=True)

        N = N_widget.value
        M = M_widget.value
        omega = omega_widget.value
        epsilon = epsilon_widget.value
        tipo_grafo = grafo_widget.value

        total = M ** N

        print("BARRIDO DE KAPPA")
        print("================")
        print(f"Total de configuraciones por kappa: {total}")
        print()

        if total > 10000:
            print("Advertencia: el barrido puede ser pesado.")
            print("Reduce N o M antes de ejecutar.")
            return

        df = barrido_kappa(
            N=N,
            M=M,
            omega=omega,
            tipo_grafo=tipo_grafo,
            epsilon=epsilon,
            T_max=150
        )

        display(df)
        plot_mapa_regimenes(df)

        if len(df) > 0:
            k_opt = int(df.loc[df["rho_sync"].idxmax(), "kappa"])
            rho_max = float(df["rho_sync"].max())
            print(f"kappa de máxima sincronización observada: {k_opt}")
            print(f"rho_sync máximo: {rho_max:.4f}")


boton_simular.on_click(on_simular_clicked)
boton_exhaustivo.on_click(on_exhaustivo_clicked)
boton_barrido.on_click(on_barrido_clicked)


# ============================================================
# 7. Panel principal
# ============================================================

panel = widgets.VBox([
    widgets.HTML("<h2>Simulador de Sincronización Discreta tipo Kuramoto sobre Grafos</h2>"),
    widgets.HTML("""
    <p>
    Modelo: <b>F<sub>κ</sub>(θ)<sub>i</sub> = θ<sub>i</sub> + ω* + σ<sub>κ</sub>(S<sub>i</sub>(θ)) mod M</b>.
    </p>
    <p>
    Convención corregida: σ<sub>κ</sub>(S)=+1 si S ≥ κ, -1 si S ≤ -κ, 0 si |S| &lt; κ.
    La subred ε-sincronizada usa d<sub>M</sub>(θ<sub>i</sub>,θ<sub>j</sub>) &lt; ε.
    </p>
    """),
    widgets.HBox([N_widget, M_widget, kappa_widget]),
    widgets.HBox([omega_widget, epsilon_widget, T_widget]),
    widgets.HBox([grafo_widget, modo_theta_widget]),
    widgets.HBox([boton_simular, boton_exhaustivo, boton_barrido]),
    salida
])


display(panel)


# ============================================================
# 8. Prueba rápida del ejemplo del paper
# ============================================================
# Si quieres probar sin widgets, ejecuta esta celda:
#
# modelo = KuramotoDiscreto(N=3, M=4, kappa=2, omega=0, tipo_grafo="Completo", epsilon=1)
# theta0 = [0, 2, 3]
# resultado = modelo.simular(theta0, T=5)
# print(resultado["trayectoria"])
# print(modelo.clasificar_atractor(theta0))
# display(modelo.tabla_trayectoria(resultado))
```

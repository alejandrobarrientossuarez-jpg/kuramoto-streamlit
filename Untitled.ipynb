```python
import itertools
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# =====================================================
# Modelo discreto de Kuramoto con umbral
# =====================================================

def dM(a, b, M):
    return ((a - b + M//2) % M) - M//2

def sigma_k(S, kappa):
    if S >= kappa:
        return 1
    elif S <= -kappa:
        return -1
    return 0

def F_kappa(theta, M, kappa, omega=0):
    N = len(theta)
    new_theta = []

    for i in range(N):
        S_i = sum(dM(theta[j], theta[i], M) for j in range(N) if j != i)
        new_theta.append((theta[i] + omega + sigma_k(S_i, kappa)) % M)

    return tuple(new_theta)

def is_synchronized(theta):
    return len(set(theta)) == 1

def classify_trajectory(theta0, M, kappa, omega=0, max_iter=100):
    seen = {}
    theta = tuple(theta0)
    path = []

    for t in range(max_iter):
        if is_synchronized(theta):
            path.append(theta)
            return "Sincronización", t, path

        if theta in seen:
            cycle_start = seen[theta]
            cycle = path[cycle_start:]

            if len(cycle) == 1:
                return "Punto fijo no trivial", t, path
            else:
                return f"Ciclo de periodo {len(cycle)}", t, path

        seen[theta] = t
        path.append(theta)
        theta = F_kappa(theta, M, kappa, omega)

    return "No determinado", max_iter, path


# =====================================================
# Grafo de acuerdo
# =====================================================

def plot_agreement_graph(theta):
    N = len(theta)
    G = nx.Graph()
    G.add_nodes_from(range(N))

    for i in range(N):
        for j in range(i + 1, N):
            if theta[i] == theta[j]:
                G.add_edge(i, j)

    pos = nx.circular_layout(G)

    labels = {i: f"{i+1}\nθ={theta[i]}" for i in range(N)}

    plt.figure(figsize=(5, 5))
    nx.draw(
        G, pos,
        with_labels=True,
        labels=labels,
        node_size=1200,
        font_size=10,
        edge_color="black"
    )
    plt.title(f"Subred de acuerdo para θ = {theta}")
    plt.show()


# =====================================================
# Simulación exhaustiva
# =====================================================

def exhaustive_simulation(N, M, omega=0):
    rows = []
    kappa_max = (N - 1) * (M // 2)

    for kappa in range(1, kappa_max + 1):
        counts = {
            "Sincronización": 0,
            "Punto fijo no trivial": 0,
            "Ciclo": 0,
            "No determinado": 0
        }

        for theta0 in itertools.product(range(M), repeat=N):
            kind, _, _ = classify_trajectory(theta0, M, kappa, omega)

            if "Sincronización" in kind:
                counts["Sincronización"] += 1
            elif "Punto fijo" in kind:
                counts["Punto fijo no trivial"] += 1
            elif "Ciclo" in kind:
                counts["Ciclo"] += 1
            else:
                counts["No determinado"] += 1

        total = M ** N

        rows.append({
            "kappa": kappa,
            "sync_%": 100 * counts["Sincronización"] / total,
            "fix_%": 100 * counts["Punto fijo no trivial"] / total,
            "cyc_%": 100 * counts["Ciclo"] / total
        })

    return pd.DataFrame(rows)


# =====================================================
# Widget principal
# =====================================================

N_widget = widgets.IntSlider(value=3, min=2, max=6, step=1, description="N")
M_widget = widgets.IntSlider(value=4, min=2, max=10, step=1, description="M")
kappa_widget = widgets.IntSlider(value=2, min=1, max=10, step=1, description="κ")
omega_widget = widgets.IntSlider(value=0, min=0, max=10, step=1, description="ω*")
theta_widget = widgets.Text(value="0,2,3", description="θ inicial")
t_widget = widgets.IntSlider(value=0, min=0, max=20, step=1, description="t")

output = widgets.Output()

def run_simulation(N, M, kappa, omega, theta_text, t):
    with output:
        clear_output()

        theta0 = tuple(int(x.strip()) % M for x in theta_text.split(","))

        if len(theta0) != N:
            print("La condición inicial debe tener exactamente N componentes.")
            return

        kind, time, path = classify_trajectory(theta0, M, kappa, omega)

        print("Resultado de la trayectoria:")
        print(f"Condición inicial: {theta0}")
        print(f"Tipo de atractor: {kind}")
        print(f"Tiempo detectado: {time}")

        df_path = pd.DataFrame(path, columns=[f"θ_{i+1}" for i in range(N)])
        df_path.insert(0, "t", range(len(df_path)))

        display(df_path)

        t = min(t, len(path) - 1)

        print(f"Subred de acuerdo en t = {t}")
        plot_agreement_graph(path[t])

        print("Simulación exhaustiva para todos los valores de κ:")
        df_exhaustive = exhaustive_simulation(N, M, omega)
        display(df_exhaustive)

        plt.figure(figsize=(7, 4))
        plt.plot(df_exhaustive["kappa"], df_exhaustive["sync_%"], marker="o", label="Sincronización")
        plt.plot(df_exhaustive["kappa"], df_exhaustive["fix_%"], marker="o", label="Puntos fijos no triviales")
        plt.plot(df_exhaustive["kappa"], df_exhaustive["cyc_%"], marker="o", label="Ciclos")
        plt.xlabel("κ")
        plt.ylabel("Porcentaje")
        plt.title(f"Regímenes dinámicos para K_{N}, M={M}")
        plt.legend()
        plt.grid(True)
        plt.show()

ui = widgets.VBox([
    widgets.HBox([N_widget, M_widget]),
    widgets.HBox([kappa_widget, omega_widget]),
    theta_widget,
    t_widget
])

interactive_output = widgets.interactive_output(
    run_simulation,
    {
        "N": N_widget,
        "M": M_widget,
        "kappa": kappa_widget,
        "omega": omega_widget,
        "theta_text": theta_widget,
        "t": t_widget
    }
)

display(ui, output, interactive_output)
```

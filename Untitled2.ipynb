```python
import requests
import pandas as pd
import os

# =====================================================
# TOKEN INEGI
# =====================================================

TOKEN = "2c63db48-9a6a-4468-be5b-8ab85da04eb6"

# =====================================================
# URL API INEGI
# =====================================================

url = (
    "https://www.inegi.org.mx/app/api/indicadores/"
    "desarrolladores/jsonxml/"
    "INDICATOR/702139,444557/"
    "es/00/false/BIE-BISE/2.0/"
    f"{TOKEN}?type=json"
)

print("Consultando API INEGI...")
print(url)

# =====================================================
# CONSULTA
# =====================================================

response = requests.get(url)

# =====================================================
# VALIDACIÓN
# =====================================================

if response.status_code == 200:

    data = response.json()

    resultados = []

    # =====================================================
    # RECORRER CADA INDICADOR
    # =====================================================

    for serie in data["Series"]:

        clave = serie.get("INDICATOR", "SinClave")
        nombre = serie.get("TITLE", "SinNombre")

        observaciones = serie.get("OBSERVATIONS", [])

        for obs in observaciones:

            resultados.append({
                "Indicador": clave,
                "Nombre": nombre,
                "Fecha": obs.get("TIME_PERIOD"),
                "Valor": obs.get("OBS_VALUE")
            })

    # =====================================================
    # DATAFRAME
    # =====================================================

    df = pd.DataFrame(resultados)

    # Convertir valores a numérico
    df["Valor"] = pd.to_numeric(df["Valor"], errors="coerce")

    # =====================================================
    # EXPORTAR CSV
    # =====================================================

    archivo_csv = "series_inegi.csv"

    df.to_csv(
        archivo_csv,
        index=False,
        encoding="utf-8-sig"
    )

    # =====================================================
    # RESULTADOS
    # =====================================================

    print("\nCSV generado correctamente.")
    print("\nPrimeras filas:")
    print(df.head())

    print("\nRuta del archivo:")
    print(os.path.abspath(archivo_csv))

else:

    print("\nERROR EN LA CONSULTA")

    print("Código:", response.status_code)

    print("\nRespuesta del servidor:")
    print(response.text)
```

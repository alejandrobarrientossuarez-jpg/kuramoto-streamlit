```python
import numpy as np
import networkx as nx


class KuramotoDiscreto:
    def __init__(self, N=3, M=4, kappa=2, omega=0, tipo_grafo="completo"):
        self.N = N
        self.M = M
        self.kappa = kappa
        self.omega = omega
        self.tipo_grafo = tipo_grafo
        self.G = self.crear_grafo()

    def crear_grafo(self):
        if self.tipo_grafo == "completo":
            return nx.complete_graph(self.N)
        elif self.tipo_grafo == "camino":
            return nx.path_graph(self.N)
        elif self.tipo_grafo == "ciclo":
            return nx.cycle_graph(self.N)
        elif self.tipo_grafo == "estrella":
            return nx.star_graph(self.N - 1)
        else:
            raise ValueError("Tipo de grafo no reconocido.")

    def dM(self, a, b):
        """
        Diferencia angular con signo:
        d_M(a,b) = ((a-b+floor(M/2)) mod M) - floor(M/2)
        """
        return ((a - b + self.M // 2) % self.M) - self.M // 2

    def sigma_kappa(self, S):
        """
        Función signo umbralizada.
        """
        if S >= self.kappa:
            return 1
        elif S <= -self.kappa:
            return -1
        else:
            return 0

    def S_i(self, theta, i):
        """
        Suma de acoplamiento del vértice i.
        """
        suma = 0
        for j in self.G.neighbors(i):
            suma += self.dM(theta[j], theta[i])
        return suma

    def paso(self, theta):
        """
        Aplica F_kappa a una configuración theta.
        """
        theta = np.array(theta, dtype=int)
        nuevo = np.zeros_like(theta)

        for i in range(self.N):
            S = self.S_i(theta, i)
            nuevo[i] = (theta[i] + self.omega + self.sigma_kappa(S)) % self.M

        return nuevo

    def simular(self, theta0, T=10):
        """
        Simula la trayectoria theta^0, theta^1, ..., theta^T.
        """
        trayectoria = [np.array(theta0, dtype=int)]

        theta = np.array(theta0, dtype=int)

        for _ in range(T):
            theta = self.paso(theta)
            trayectoria.append(theta.copy())

        return trayectoria

    def esta_sincronizado(self, theta):
        """
        Verifica si theta pertenece a la diagonal Delta_M.
        """
        return np.all(theta == theta[0])
```

```python
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.patches import FancyBboxPatch
import statsmodels.api as sm
from statsmodels.stats.stattools import durbin_watson
from statsmodels.tsa.stattools import adfuller
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# ── Paleta de colores institucional ──────────────────────────────────────────
C1  = '#1B3A6B'   # azul marino
C2  = '#C8372D'   # rojo
C3  = '#2E7D32'   # verde
C4  = '#F57F17'   # naranja
C5  = '#6A1B9A'   # morado
BG  = '#F4F6F9'
CARD= '#FFFFFF'

# ── Cargar y preparar datos ───────────────────────────────────────────────────
df = pd.read_excel(r'C:\Users\USUARIO\Downloads\Base_de_datos_EQUIPO_2.xlsx',
                   sheet_name='Base de datos')


months_es = {'Ene':1,'Feb':2,'Mar':3,'Abr':4,'May':5,'Jun':6,
             'Jul':7,'Ago':8,'Sep':9,'Oct':10,'Nov':11,'Dic':12}

def parse_fecha(s):
    m, y = s.split()
    return pd.Timestamp(year=int(y), month=months_es[m], day=1)

df['fecha'] = df['Fecha'].apply(parse_fecha)
df = df.sort_values(['Entidad','fecha']).reset_index(drop=True)

# ── Variables por entidad ─────────────────────────────────────────────────────
entidades = ['San Luis Potosi','Aguascalientes','Queretaro','Guanajuato']
colores_ent = {e:c for e,c in zip(entidades,[C1,C2,C3,C4])}

def prep(sub):
    sub = sub.copy().sort_values('fecha').reset_index(drop=True)
    # Salario real = Remuneraciones / Personal / INPC * 100
    sub['salario_nominal'] = sub['Remuneraciones pagadas '] / sub['Personal ocupado total ']
    sub['salario_real']    = (sub['salario_nominal'] / sub['INPC (mensual)']) * 100
    # Índice de personal (base 2018 = 100)
    base = sub.loc[sub['fecha'].dt.year==2018,'Personal ocupado total '].mean()
    sub['ind_personal'] = (sub['Personal ocupado total '] / base) * 100
    # Crecimiento anual personal ocupado
    sub['crec_personal'] = sub['Personal ocupado total '].pct_change(12) * 100
    # Log-variables para regresión
    sub['ln_salario_real']  = np.log(sub['salario_real'])
    sub['ln_personal']      = np.log(sub['Personal ocupado total '])
    sub['tendencia']        = np.arange(len(sub))
    return sub

datos = {e: prep(df[df['Entidad']==e]) for e in entidades}
slp   = datos['San Luis Potosi']

# ─────────────────────────────────────────────────────────────────────────────
# REGRESIÓN ECONOMÉTRICA  (OLS con corrección Newey-West)
#   ln(Salario_Real_t) = β0 + β1·ln(Personal_t) + β2·Tendencia + ε
#   Hipótesis: β1 > 0  → más empleo industrial → mayor presión salarial
# ─────────────────────────────────────────────────────────────────────────────
reg = slp.dropna(subset=['ln_salario_real','ln_personal']).copy()
X   = sm.add_constant(reg[['ln_personal','tendencia']])
y   = reg['ln_salario_real']

modelo_ols   = sm.OLS(y, X).fit()
modelo_nw    = modelo_ols.get_robustcov_results(cov_type='HAC', maxlags=4)
dw_stat      = durbin_watson(modelo_ols.resid)
y_pred       = modelo_ols.fittedvalues
residuos     = modelo_ols.resid

# ADF sobre residuos (cointegración)
adf_res      = adfuller(residuos.values, autolag='AIC')

# param names: const, ln_personal, tendencia  (indices 0,1,2)
params_arr   = np.array(modelo_nw.params)
bse_arr      = np.array(modelo_nw.bse)
tval_arr     = np.array(modelo_nw.tvalues)
pval_arr     = np.array(modelo_nw.pvalues)
param_names  = X.columns.tolist()   # ['const','ln_personal','tendencia']

def gp(name):  # get param index
    return param_names.index(name)

beta0  = params_arr[gp('const')]
beta1  = params_arr[gp('ln_personal')]
beta2  = params_arr[gp('tendencia')]
pval_b1 = pval_arr[gp('ln_personal')]
elasticidad = beta1

rsq     = modelo_ols.rsquared
rsq_adj = modelo_ols.rsquared_adj
fval    = modelo_ols.fvalue
fp      = modelo_ols.f_pvalue
nobs    = int(modelo_ols.nobs)

# ─────────────────────────────────────────────────────────────────────────────
# ───────────────── FIGURA PRINCIPAL (Dashboard) ──────────────────────────────
# ─────────────────────────────────────────────────────────────────────────────
fig = plt.figure(figsize=(22, 28), facecolor=BG)
fig.suptitle('ANÁLISIS ECONOMÉTRICO DEL MERCADO LABORAL MANUFACTURERO\nSan Luis Potosí y Región Centro-Norte  |  Ene 2018 – Dic 2025',
             fontsize=17, fontweight='bold', color=C1, y=0.99, ha='center')

gs = gridspec.GridSpec(5, 3, figure=fig,
                       hspace=0.55, wspace=0.38,
                       top=0.96, bottom=0.04,
                       left=0.06, right=0.97)

# ── Funciones auxiliares ──────────────────────────────────────────────────────
def card_bg(ax, color=CARD, alpha=0.92):
    ax.set_facecolor(color)
    for sp in ax.spines.values():
        sp.set_color('#D0D7E3')
        sp.set_linewidth(0.8)

def titulo(ax, txt, sub=''):
    ax.set_title(f'$\\bf{{{txt}}}$' + (f'\n{sub}' if sub else ''),
                 fontsize=10, color=C1, pad=8, loc='left')

# ─── KPI Cards (fila 0) ───────────────────────────────────────────────────────
kpis = [
    ('Personal Ocupado\n(Dic 2025)', f"{slp['Personal ocupado total '].iloc[-1]:,.0f}",
     f"Var. anual: {slp['Personal ocupado total '].pct_change(12).iloc[-1]*100:+.1f}%", C1),
    ('Salario Real\n(Dic 2025, índice)', f"{slp['salario_real'].iloc[-1]:,.1f}",
     f"Var. anual: {(slp['salario_real'].iloc[-1]/slp['salario_real'].iloc[-13]-1)*100:+.1f}%", C3),
    ('Elasticidad\nSalarial Estimada', f"{elasticidad:.4f}",
     f"p-valor: {pval_b1:.4f}  |  R²={modelo_nw.rsquared:.3f}", C2 if pval_b1<0.05 else C4),
]
for col, (lbl, val, sub, color) in enumerate(kpis):
    ax = fig.add_subplot(gs[0, col])
    ax.set_xlim(0,1); ax.set_ylim(0,1); ax.axis('off')
    rect = FancyBboxPatch((0.04,0.06), 0.92, 0.88,
                          boxstyle="round,pad=0.04",
                          facecolor=color, edgecolor='none', alpha=0.92)
    ax.add_patch(rect)
    ax.text(0.50, 0.72, lbl, ha='center', va='center',
            fontsize=10, color='white', fontweight='bold')
    ax.text(0.50, 0.44, val, ha='center', va='center',
            fontsize=19, color='white', fontweight='bold')
    ax.text(0.50, 0.18, sub, ha='center', va='center',
            fontsize=8.5, color='#FFFFFFCC')

# ─── G1: Personal Ocupado – 4 entidades ──────────────────────────────────────
ax1 = fig.add_subplot(gs[1, :2])
card_bg(ax1)
for e in entidades:
    d = datos[e]
    ax1.plot(d['fecha'], d['Personal ocupado total '],
             color=colores_ent[e], lw=2, label=e)
ax1.axvspan(pd.Timestamp('2020-03-01'), pd.Timestamp('2020-09-01'),
            color='gray', alpha=0.15, label='COVID-19')
ax1.set_ylabel('Personas', fontsize=9)
ax1.legend(fontsize=8, loc='upper left', framealpha=0.7)
ax1.yaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f'{x/1000:.0f}k'))
titulo(ax1,'Personal Ocupado Total – Sector Manufacturero',
       'Miles de trabajadores | Comparativo regional')

# ─── G2: Salario Real SLP ────────────────────────────────────────────────────
ax2 = fig.add_subplot(gs[1, 2])
card_bg(ax2)
ax2.fill_between(slp['fecha'], slp['salario_real'], alpha=0.25, color=C1)
ax2.plot(slp['fecha'], slp['salario_real'], color=C1, lw=2)
ax2.set_ylabel('Índice (base 2018)', fontsize=9)
titulo(ax2,'Salario Real SLP','Remuneración/empleado deflactada por INPC')

# ─── G3: Scatter + ajuste OLS ────────────────────────────────────────────────
ax3 = fig.add_subplot(gs[2, 0])
card_bg(ax3)
sc = ax3.scatter(reg['ln_personal'], reg['ln_salario_real'],
                 c=np.arange(len(reg)), cmap='Blues', s=25, alpha=0.8, zorder=3)
xr = np.linspace(reg['ln_personal'].min(), reg['ln_personal'].max(), 100)
Xr = np.column_stack([np.ones(100), xr, np.linspace(0, len(reg)-1, 100)])
yr = Xr @ params_arr
ax3.plot(xr, yr, color=C2, lw=2, label=f'OLS  β₁={beta1:.3f}')
ax3.set_xlabel('ln(Personal Ocupado)', fontsize=9)
ax3.set_ylabel('ln(Salario Real)', fontsize=9)
ax3.legend(fontsize=8)
titulo(ax3,'Dispersión Log-Log','Empleo vs Salario Real')

# ─── G4: Residuos en el tiempo ───────────────────────────────────────────────
ax4 = fig.add_subplot(gs[2, 1])
card_bg(ax4)
ax4.bar(reg['fecha'], residuos, color=np.where(residuos>=0, C3, C2), alpha=0.7, width=20)
ax4.axhline(0, color='black', lw=0.8)
ax4.set_ylabel('Residuo', fontsize=9)
titulo(ax4,'Residuos del Modelo OLS',f'Durbin-Watson = {dw_stat:.3f}')

# ─── G5: QQ-plot residuos ────────────────────────────────────────────────────
ax5 = fig.add_subplot(gs[2, 2])
card_bg(ax5)
(osm, osr), (slope, intercept, r) = stats.probplot(residuos, dist='norm')
ax5.scatter(osm, osr, color=C1, s=20, alpha=0.7)
ax5.plot(osm, slope*np.array(osm)+intercept, color=C2, lw=1.5)
ax5.set_xlabel('Cuantiles teóricos', fontsize=9)
ax5.set_ylabel('Cuantiles observados', fontsize=9)
titulo(ax5,'Q-Q Plot Residuos',f'Normalidad: r={r:.3f}')

# ─── G6: Índice personal (base 100, 4 entidades) ─────────────────────────────
ax6 = fig.add_subplot(gs[3, :2])
card_bg(ax6)
for e in entidades:
    d = datos[e]
    ax6.plot(d['fecha'], d['ind_personal'], color=colores_ent[e], lw=2, label=e)
ax6.axhline(100, color='gray', lw=0.8, ls='--')
ax6.axvspan(pd.Timestamp('2020-03-01'), pd.Timestamp('2020-09-01'),
            color='gray', alpha=0.15)
ax6.set_ylabel('Índice (Ene 2018 = 100)', fontsize=9)
ax6.legend(fontsize=8)
titulo(ax6,'Índice de Crecimiento Industrial (Personal Ocupado)',
       'Base Enero 2018 = 100 | Comparativo regional')

# ─── G7: Crecimiento anual SLP ───────────────────────────────────────────────
ax7 = fig.add_subplot(gs[3, 2])
card_bg(ax7)
crec = slp['crec_personal'].dropna()
ax7.bar(slp.loc[crec.index,'fecha'], crec,
        color=np.where(crec>=0, C3, C2), alpha=0.8, width=20)
ax7.axhline(0, color='black', lw=0.8)
ax7.set_ylabel('% var. anual', fontsize=9)
titulo(ax7,'Crecimiento Anual Empleo SLP','Variación % interanual')

# ─── G8: Tabla de resultados econométricos ───────────────────────────────────
ax8 = fig.add_subplot(gs[4, :])
ax8.axis('off')
card_bg(ax8)

# Encabezado tabla
ax8.text(0.5, 0.95,
         'RESULTADOS DE LA REGRESIÓN ECONOMÉTRICA (OLS – Errores Estándar HAC Newey-West)',
         ha='center', va='top', fontsize=11, fontweight='bold', color=C1,
         transform=ax8.transAxes)

i0,i1,i2 = gp('const'), gp('ln_personal'), gp('tendencia')
tabla_data = [
    ['Variable','Coeficiente','Error Est. (NW)','t-estadístico','p-valor','Interpretación'],
    ['Constante (β₀)',
     f"{params_arr[i0]:.4f}", f"{bse_arr[i0]:.4f}", f"{tval_arr[i0]:.3f}", f"{pval_arr[i0]:.4f}",
     'Nivel base del salario real'],
    ['ln(Personal Ocu.) (β₁)',
     f"{params_arr[i1]:.4f}", f"{bse_arr[i1]:.4f}", f"{tval_arr[i1]:.3f}",
     f"{'***' if pval_b1<0.01 else '**' if pval_b1<0.05 else '*' if pval_b1<0.1 else ''} {pval_b1:.4f}",
     f'Elasticidad salarial: +1% empleo → +{beta1:.2f}% salario real'],
    ['Tendencia (β₂)',
     f"{params_arr[i2]:.4f}", f"{bse_arr[i2]:.4f}", f"{tval_arr[i2]:.3f}", f"{pval_arr[i2]:.4f}",
     'Tendencia secular del salario real'],
    ['—','—','—','—','—','—'],
    ['R²',f"{rsq:.4f}",'','','','Bondad de ajuste del modelo'],
    ['R² ajustado',f"{rsq_adj:.4f}",'','','','Ajustado por grados de libertad'],
    ['F-estadístico',f"{fval:.3f}",'','p(F)=',f"{fp:.6f}",'Significancia global del modelo'],
    ['Durbin-Watson',f"{dw_stat:.4f}",'','','','Autocorrelación residual (≈2 = sin autocor.)'],
    ['ADF residuos (p)',f"{adf_res[1]:.4f}",'','','',
     'Estacionariedad residuos (p<0.05 → cointegración)'],
    ['Obs.',f"{nobs}",'','','','Datos mensuales Ene 2018 – Dic 2025'],
]

col_x = [0.01, 0.16, 0.30, 0.42, 0.55, 0.66]
col_w = [15, 14, 14, 13, 13, 34]
row_h = 0.088
y0    = 0.88

for r_idx, row in enumerate(tabla_data):
    y_pos = y0 - r_idx * row_h
    bg_color = '#E8EDF6' if r_idx == 0 else (BG if r_idx % 2 == 0 else CARD)
    if row[0] == '—':
        ax8.plot([0.01, 0.99], [y_pos + row_h*0.3, y_pos + row_h*0.3],
                 color='#C0C8D8', lw=0.6, transform=ax8.transAxes, clip_on=False)
        continue
    rect = FancyBboxPatch((0.005, y_pos - row_h*0.55), 0.99, row_h*0.95,
                          boxstyle="square,pad=0",
                          facecolor=bg_color, edgecolor='none',
                          transform=ax8.transAxes, clip_on=False)
    ax8.add_patch(rect)
    fw = 'bold' if r_idx == 0 else 'normal'
    fc = 'white' if r_idx == 0 else 'black'
    if r_idx == 0:
        rect.set_facecolor(C1)
    for c_idx, (cell, cx, cw) in enumerate(zip(row, col_x, col_w)):
        ax8.text(cx + 0.005, y_pos - row_h*0.05, cell,
                 ha='left', va='center', fontsize=8.2,
                 fontweight=fw, color=fc,
                 transform=ax8.transAxes)

# Nota metodológica
ax8.text(0.5, -0.05,
         'Modelo: ln(Salario_Real)ₜ = β₀ + β₁·ln(Personal_Ocu)ₜ + β₂·Tendencia + εₜ  '
         '|  Fuente: INEGI-EMIM & Banco de México (INPC)  |  Elaboración propia',
         ha='center', va='bottom', fontsize=8, color='#555555',
         style='italic', transform=ax8.transAxes)

plt.savefig('/mnt/user-data/outputs/dashboard_slp.png',
            dpi=160, bbox_inches='tight', facecolor=BG)
plt.close()
print("Dashboard guardado.")

# ─────────────────────────────────────────────────────────────────────────────
# SEGUNDA FIGURA: Análisis profundo de la pregunta central
# ─────────────────────────────────────────────────────────────────────────────
fig2, axes = plt.subplots(2, 2, figsize=(18, 12), facecolor=BG)
fig2.suptitle('¿El Crecimiento Industrial ha Presionado los Salarios Reales\n'
              'del Sector Manufacturero en San Luis Potosí?',
              fontsize=15, fontweight='bold', color=C1, y=1.01)

# Eje dual: personal + salario real
ax = axes[0,0]
card_bg(ax)
ax2r = ax.twinx()
ax.fill_between(slp['fecha'], slp['Personal ocupado total '],
                alpha=0.2, color=C1)
ax.plot(slp['fecha'], slp['Personal ocupado total '],
        color=C1, lw=2.2, label='Personal Ocupado')
ax2r.plot(slp['fecha'], slp['salario_real'],
          color=C2, lw=2.2, ls='--', label='Salario Real')
ax.set_ylabel('Personal Ocupado (personas)', color=C1, fontsize=9)
ax2r.set_ylabel('Salario Real (índice)', color=C2, fontsize=9)
lines1, labels1 = ax.get_legend_handles_labels()
lines2, labels2 = ax2r.get_legend_handles_labels()
ax.legend(lines1+lines2, labels1+labels2, fontsize=8, loc='upper left')
ax.set_title('Personal Ocupado vs Salario Real\n(Eje dual)', fontsize=10, color=C1, fontweight='bold')

# Correlación rolling 24m
ax = axes[0,1]
card_bg(ax)
roll_corr = slp['Personal ocupado total '].rolling(24).corr(slp['salario_real'])
ax.plot(slp['fecha'], roll_corr, color=C5, lw=2)
ax.axhline(0, color='gray', lw=0.8, ls='--')
ax.fill_between(slp['fecha'], roll_corr, 0,
                where=roll_corr>0, alpha=0.3, color=C3, label='Positiva')
ax.fill_between(slp['fecha'], roll_corr, 0,
                where=roll_corr<0, alpha=0.3, color=C2, label='Negativa')
ax.set_ylabel('Correlación', fontsize=9)
ax.legend(fontsize=8)
ax.set_title('Correlación Rodante (24 meses)\nPersonal Ocupado ↔ Salario Real', fontsize=10, color=C1, fontweight='bold')

# Valores ajustados vs reales
ax = axes[1,0]
card_bg(ax)
ax.plot(reg['fecha'], np.exp(y.values), color=C1, lw=2, label='Observado', alpha=0.9)
ax.plot(reg['fecha'], np.exp(y_pred.values), color=C2, lw=2, ls='--', label='Ajustado OLS', alpha=0.9)
ax.set_ylabel('Salario Real (índice)', fontsize=9)
ax.legend(fontsize=8)
ax.set_title('Salario Real: Observado vs Predicho\nModelo OLS Log-Log', fontsize=10, color=C1, fontweight='bold')

# Descomposición: aumento nominal vs INPC
ax = axes[1,1]
card_bg(ax)
base_nom  = slp['salario_nominal'].iloc[0]
base_inpc = slp['INPC (mensual)'].iloc[0]
idx_nom   = (slp['salario_nominal'] / base_nom) * 100
idx_inpc  = (slp['INPC (mensual)']  / base_inpc) * 100
ax.plot(slp['fecha'], idx_nom,  color=C1, lw=2, label='Salario Nominal (índice)')
ax.plot(slp['fecha'], idx_inpc, color=C2, lw=2, ls='--', label='INPC (índice)')
ax.fill_between(slp['fecha'], idx_nom, idx_inpc,
                where=idx_nom>=idx_inpc, alpha=0.25, color=C3, label='Ganancia real')
ax.fill_between(slp['fecha'], idx_nom, idx_inpc,
                where=idx_nom<idx_inpc,  alpha=0.25, color=C2, label='Pérdida real')
ax.axhline(100, color='gray', lw=0.6, ls=':')
ax.set_ylabel('Índice (Ene 2018 = 100)', fontsize=9)
ax.legend(fontsize=8)
ax.set_title('Salario Nominal vs Inflación (INPC)\nGanancia/Pérdida de Poder Adquisitivo', fontsize=10, color=C1, fontweight='bold')

for ax in axes.flat:
    ax.tick_params(labelsize=8)
    ax.set_facecolor(CARD)

plt.tight_layout(pad=2.5)
plt.savefig('/mnt/user-data/outputs/analisis_pregunta_central.png',
            dpi=160, bbox_inches='tight', facecolor=BG)
plt.close()
print("Segunda figura guardada.")

# ─────────────────────────────────────────────────────────────────────────────
# IMPRIMIR RESUMEN DE RESULTADOS
# ─────────────────────────────────────────────────────────────────────────────
print("\n" + "="*65)
print("  RESUMEN ECONOMÉTRICO – SAN LUIS POTOSÍ (Manufactura)")
print("="*65)
print(f"\nModelo: ln(Salario_Real) = β0 + β1·ln(Personal) + β2·t + ε")
print(f"\nβ1 (elasticidad) = {beta1:.4f}   p-valor = {pval_b1:.4f}")
sig = "SIGNIFICATIVO ***" if pval_b1<0.01 else "SIGNIFICATIVO **" if pval_b1<0.05 else "marginalmente sig." if pval_b1<0.1 else "NO significativo"
print(f"→ {sig}")
print(f"→ Un incremento de 1% en empleo industrial se asocia a un")
print(f"  cambio de {beta1:.3f}% en el salario real manufacturero.")
print(f"\nR² = {rsq:.4f}   R²aj = {rsq_adj:.4f}")
print(f"Durbin-Watson = {dw_stat:.4f}")
print(f"ADF sobre residuos: p = {adf_res[1]:.4f}")
print(f"\nRespuesta a la pregunta central:")
if pval_b1 < 0.05 and beta1 > 0:
    print("  SÍ: El crecimiento industrial (mayor empleo manufacturero)")
    print("  ha ejercido presión POSITIVA sobre los salarios reales en SLP.")
elif pval_b1 < 0.05 and beta1 < 0:
    print("  PARADOJA: El crecimiento industrial está asociado a MENOR")
    print("  salario real (posible efecto composición o exceso de oferta).")
else:
    print("  EVIDENCIA DÉBIL: No se puede rechazar β1=0 al 5%.")
print("="*65)
```

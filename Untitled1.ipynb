```python
# --- SCRIPT PARA CREACIÓN DE BASE DE DATOS ESTATAL (INEGI) ---

import requests
import pandas as pd

# 1. Configuración inicial
# ---------------------------------------------------------

MI_TOKEN = "2c63db48-9a6a-4468-be5b-8ab85da04eb6".strip()

claves_region = {
    "Aguascalientes": "789780",
    "Guanajuato": "789791",
    "Jalisco": "789794",
    "Queretaro": "789802",
    "San Luis Potosi": "789804"
}

# 2. Función de extracción y limpieza
# ---------------------------------------------------------

def obtener_datos_inegi(indicador_id, nombre_estado):
    indicador_id = str(indicador_id).strip()
    nombre_estado = str(nombre_estado).strip()

    endpoint = (
        "https://www.inegi.org.mx/app/api/indicadores/desarrolladores/jsonxml/"
        f"INDICATOR/{indicador_id}/es/99/false/BISE/2.0/{MI_TOKEN}?type=json"
    )

    print(f"Consultando: {nombre_estado}")
    print(endpoint)

    respuesta = requests.get(endpoint)

    if respuesta.status_code != 200:
        print(f"Error al conectar con {nombre_estado} - Código: {respuesta.status_code}")
        print("Respuesta de INEGI:")
        print(respuesta.text)
        return None

    data = respuesta.json()

    if "Series" not in data or len(data["Series"]) == 0:
        print(f"No se encontraron series para {nombre_estado}")
        return None

    observaciones = data["Series"][0].get("OBSERVATIONS", [])

    if not observaciones:
        print(f"No se encontraron observaciones para {nombre_estado}")
        return None

    df = pd.DataFrame(observaciones)

    df_clean = df[["TIME_PERIOD", "OBS_VALUE"]].copy()

    df_clean = df_clean.rename(columns={
        "TIME_PERIOD": "Periodo_Raw",
        "OBS_VALUE": "Valor"
    })

    df_clean["Estado"] = nombre_estado
    df_clean["Valor"] = pd.to_numeric(df_clean["Valor"], errors="coerce")

    df_clean["Fecha"] = pd.to_datetime(
        df_clean["Periodo_Raw"].str.replace("/", "-", regex=False) + "-01",
        errors="coerce"
    )

    df_clean["Año"] = df_clean["Fecha"].dt.year
    df_clean["Trimestre"] = df_clean["Fecha"].dt.quarter

    df_clean = df_clean[[
        "Estado",
        "Fecha",
        "Año",
        "Trimestre",
        "Valor"
    ]]

    return df_clean

# 3. Ejecución y creación de la base maestra
# ---------------------------------------------------------

lista_dfs = []

for estado, indicador in claves_region.items():
    df_estado = obtener_datos_inegi(indicador, estado)

    if df_estado is not None:
        lista_dfs.append(df_estado)

if lista_dfs:
    df_manufactura = pd.concat(lista_dfs, ignore_index=True)
else:
    df_manufactura = pd.DataFrame()

# 4. Verificación de resultados
# ---------------------------------------------------------

print("\nEstructura de la base de datos creada:")
print(df_manufactura.info())

print("\nPrimeros registros:")
print(df_manufactura.head())

print("\nÚltimos registros:")
print(df_manufactura.tail())

# 5. Guardar resultado
# ---------------------------------------------------------

df_manufactura.to_csv(
    "base_empleo_manufacturero_bajio.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nArchivo guardado exitosamente como: base_empleo_manufacturero_bajio.csv")
```

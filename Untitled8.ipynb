```python
# ============================================================
# Simulador Discreto tipo Kuramoto sobre Grafos  —  versión mínima
#
# Modelo (paper):
#   F_kappa(theta)_i = theta_i + omega* + sigma_kappa(S_i(theta))  mod M
#   S_i(theta)       = sum_{j in N(i)} d_M(theta_j, theta_i)
#   d_M(a,b)         = ((a-b + floor(M/2)) mod M) - floor(M/2)
#   sigma_kappa(S)   = +1 si S>=kappa,  -1 si S<=-kappa,  0 si |S|<kappa
# ============================================================

import numpy as np
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output
from itertools import product


# ============================================================
# Núcleo del modelo
# ============================================================

class KuramotoDiscreto:

    def __init__(self, N, M, kappa, omega=0, tipo_grafo="Completo"):
        self.N, self.M, self.kappa, self.omega = int(N), int(M), int(kappa), int(omega)
        self.G   = self._grafo(tipo_grafo)
        self.pos = nx.spring_layout(self.G, seed=7)

    def _grafo(self, tipo):
        return {
            "Completo": nx.complete_graph(self.N),
            "Camino":   nx.path_graph(self.N),
            "Ciclo":    nx.cycle_graph(self.N),
            "Estrella": nx.star_graph(self.N - 1),
        }[tipo]

    # --- distancias ---
    def _dM(self, a, b):
        """d_M(a,b) con signo, rango [-M//2, M//2]."""
        return ((int(a) - int(b) + self.M // 2) % self.M) - self.M // 2

    def _dM_abs(self, a, b):
        """Distancia circular sin signo."""
        d = abs(int(a) - int(b))
        return min(d, self.M - d)

    # --- acoplamiento ---
    def _S(self, theta):
        return np.array([
            sum(self._dM(theta[j], theta[i]) for j in self.G.neighbors(i))
            for i in range(self.N)
        ], dtype=int)

    def _sigma(self, S):
        return np.where(S >= self.kappa, 1, np.where(S <= -self.kappa, -1, 0))

    # --- un paso ---
    def paso(self, theta):
        S = self._S(theta)
        return (theta + self.omega + self._sigma(S)) % self.M

    # --- trayectoria ---
    def simular(self, theta0, T):
        theta = np.array(theta0, dtype=int) % self.M
        tray  = [theta.copy()]
        r     = [self._r(theta)]
        for _ in range(T):
            theta = self.paso(theta)
            tray.append(theta.copy())
            r.append(self._r(theta))
        return np.array(tray), np.array(r)

    # --- parámetro de orden ---
    def _r(self, theta):
        return float(np.abs(np.mean(np.exp(2j * np.pi * theta / self.M))))

    # --- sincronización ---
    def sincronizado(self, theta):
        return bool(np.all(theta == theta[0]))

    # --- clasificador de atractor ---
    def clasificar(self, theta0, T_max=300):
        theta  = np.array(theta0, dtype=int) % self.M
        vistos = {}
        for t in range(T_max + 1):
            key = tuple(theta)
            if self.sincronizado(theta):
                return "sync", t
            if key in vistos:
                periodo = t - vistos[key]
                return ("fijo" if periodo == 1 else "ciclo"), t
            vistos[key] = t
            theta = self.paso(theta)
        return "transitorio", T_max


# ============================================================
# Exploración exhaustiva (solo lo esencial)
# ============================================================

def explorar(N, M, kappa, omega, tipo_grafo, T_max=150):
    modelo = KuramotoDiscreto(N, M, kappa, omega, tipo_grafo)
    total  = M ** N
    conteo = {"sync": 0, "fijo": 0, "ciclo": 0, "transitorio": 0}
    t_sync = []

    for theta0 in product(range(M), repeat=N):
        tipo, t = modelo.clasificar(theta0, T_max)
        conteo[tipo] += 1
        if tipo == "sync":
            t_sync.append(t)

    return {
        "total":      total,
        "conteo":     conteo,
        "rho":        {k: v / total for k, v in conteo.items()},
        "t_sync":     t_sync,
    }


# ============================================================
# Conclusión
# ============================================================

def conclusion_sim(clasif, t, theta_final, r0, rT):
    marcas = {"sync": "✔  SINCRONIZÓ", "fijo": "✘  NO sincronizó (fijo)",
              "ciclo": "✘  NO sincronizó (ciclo)", "transitorio": "?  Indeterminado"}
    lineas = [
        "─" * 44,
        f"  CONCLUSIÓN",
        "─" * 44,
        f"  {marcas[clasif]}",
        f"  Tiempo detectado : t = {t}",
        f"  Estado final     : {tuple(theta_final)}",
        f"  r(0) = {r0:.3f}  →  r(T) = {rT:.3f}",
    ]
    if clasif == "transitorio":
        lineas.append("  Aumenta T para continuar la evolución.")
    lineas.append("─" * 44)
    print("\n".join(lineas))


def conclusion_exhaustiva(res):
    rho   = res["rho"]
    t_s   = res["t_sync"]
    dominante = max(rho, key=rho.get)
    lineas = [
        "─" * 44,
        "  CONCLUSIÓN  (exploración exhaustiva)",
        "─" * 44,
        f"  Total configuraciones : {res['total']}",
        f"  sync       : {res['conteo']['sync']:>5}  ({rho['sync']*100:5.1f} %)",
        f"  fijo       : {res['conteo']['fijo']:>5}  ({rho['fijo']*100:5.1f} %)",
        f"  ciclo      : {res['conteo']['ciclo']:>5}  ({rho['ciclo']*100:5.1f} %)",
        f"  transitorio: {res['conteo']['transitorio']:>5}  ({rho['transitorio']*100:5.1f} %)",
        "─" * 44,
    ]
    if rho["sync"] >= 0.5:
        lineas.append(f"  ✔  MAYORÍA sincroniza  ({rho['sync']*100:.1f} %)")
    elif rho["sync"] > 0:
        lineas.append(f"  ~  MINORÍA sincroniza  ({rho['sync']*100:.1f} %)  —  domina: {dominante}")
    else:
        lineas.append(f"  ✘  NINGUNA condición inicial sincroniza  —  domina: {dominante}")
    if t_s:
        lineas.append(f"  Tiempo sync: media={np.mean(t_s):.1f}  máx={max(t_s)}")
    lineas.append("─" * 44)
    print("\n".join(lineas))


# ============================================================
# Visualizaciones (mínimas)
# ============================================================

def plot_heatmap(tray, M):
    fig, ax = plt.subplots(figsize=(8, 3))
    im = ax.imshow(tray.T, aspect="auto", interpolation="nearest", vmin=0, vmax=M - 1)
    ax.set_xlabel("t")
    ax.set_ylabel("vértice")
    ax.set_title("Evolución de fases")
    plt.colorbar(im, ax=ax, label="fase")
    plt.tight_layout(); plt.show()


def plot_r(r):
    fig, ax = plt.subplots(figsize=(8, 2.5))
    ax.plot(r, marker="o", markersize=3)
    ax.set_ylim(-0.05, 1.05)
    ax.set_xlabel("t"); ax.set_ylabel("r(t)")
    ax.set_title("Parámetro de orden")
    ax.grid(True, alpha=0.3)
    plt.tight_layout(); plt.show()


def plot_grafo(modelo, theta, titulo):
    fig, ax = plt.subplots(figsize=(4, 4))
    # aristas normales
    nx.draw_networkx_edges(modelo.G, modelo.pos, ax=ax, alpha=0.2)
    # aristas en acuerdo (distancia 0)
    acuerdo = [(i, j) for i, j in modelo.G.edges()
               if modelo._dM_abs(theta[i], theta[j]) == 0]
    if acuerdo:
        nx.draw_networkx_edges(modelo.G, modelo.pos, edgelist=acuerdo,
                               ax=ax, width=2.5, edge_color="tab:blue")
    nodes = nx.draw_networkx_nodes(
        modelo.G, modelo.pos,
        node_color=theta.astype(float),
        cmap=plt.cm.viridis, vmin=0, vmax=modelo.M - 1,
        node_size=500, ax=ax,
    )
    nx.draw_networkx_labels(modelo.G, modelo.pos, ax=ax, font_size=9)
    plt.colorbar(nodes, ax=ax, label="fase")
    ax.set_title(titulo); ax.axis("off")
    plt.tight_layout(); plt.show()


# ============================================================
# Widgets
# ============================================================

N_w     = widgets.IntSlider(value=3, min=2, max=15, step=1, description="N",
                             style={"description_width": "50px"}, layout=widgets.Layout(width="300px"))
M_w     = widgets.IntSlider(value=4, min=2, max=12, step=1, description="M",
                             style={"description_width": "50px"}, layout=widgets.Layout(width="300px"))
kappa_w = widgets.IntSlider(value=2, min=1, max=20, step=1, description="κ",
                             style={"description_width": "50px"}, layout=widgets.Layout(width="300px"))
omega_w = widgets.IntSlider(value=0, min=0, max=11, step=1, description="ω*",
                             style={"description_width": "50px"}, layout=widgets.Layout(width="300px"))
T_w     = widgets.IntSlider(value=20, min=1, max=100, step=1, description="T",
                             style={"description_width": "50px"}, layout=widgets.Layout(width="300px"))

grafo_w = widgets.Dropdown(options=["Completo", "Camino", "Ciclo", "Estrella"],
                            value="Completo", description="Grafo",
                            style={"description_width": "50px"})

theta0_w = widgets.Dropdown(options=["Ejemplo paper", "Aleatoria", "Uniforme"],
                             value="Ejemplo paper", description="θ₀",
                             style={"description_width": "50px"})

btn_sim = widgets.Button(description="▶  Simular",       button_style="success")
btn_exp = widgets.Button(description="⊞  Exhaustiva",    button_style="warning")
salida  = widgets.Output()


# rango dinámico de kappa y omega
def _actualizar(*_):
    N, M, tipo = N_w.value, M_w.value, grafo_w.value
    omega_w.max = M - 1
    ref = KuramotoDiscreto(N, M, 1, tipo_grafo=tipo)
    dmax = max(dict(ref.G.degree()).values())
    kmax = max(1, dmax * (M // 2))
    kappa_w.value = min(kappa_w.value, kmax)
    kappa_w.max   = kmax

for w in [N_w, M_w, grafo_w]:
    w.observe(_actualizar, names="value")
_actualizar()


def _theta0(N, M, modo):
    if modo == "Ejemplo paper":
        base = [0, M // 2, M - 1]
        return np.array((base + [0] * N)[:N], dtype=int) % M
    if modo == "Uniforme":
        return np.array([i % M for i in range(N)], dtype=int)
    return np.random.randint(0, M, size=N)


# ── Simulación individual ──────────────────────────────────
def on_sim(b):
    with salida:
        clear_output(wait=True)
        N, M, kappa = N_w.value, M_w.value, kappa_w.value
        omega, T    = omega_w.value, T_w.value
        tipo        = grafo_w.value

        modelo  = KuramotoDiscreto(N, M, kappa, omega, tipo)
        theta0  = _theta0(N, M, theta0_w.value)
        tray, r = modelo.simular(theta0, T)
        clasif, t_det = modelo.clasificar(theta0, T_max=max(300, T))

        print(f"Grafo={tipo}  N={N}  M={M}  κ={kappa}  ω*={omega}  T={T}")
        print(f"θ₀ = {tuple(theta0)}")
        print()

        plot_heatmap(tray, M)
        plot_r(r)
        plot_grafo(modelo, tray[0],  f"t = 0")
        plot_grafo(modelo, tray[-1], f"t = {T}")

        # tabla compacta
        filas = [{"t": t,
                  "θ(t)": tuple(tray[t]),
                  "r(t)": round(r[t], 3)}
                 for t in range(len(tray))]
        display(pd.DataFrame(filas))

        conclusion_sim(clasif, t_det, tray[-1], r[0], r[-1])


# ── Exploración exhaustiva ─────────────────────────────────
def on_exp(b):
    with salida:
        clear_output(wait=True)
        N, M, kappa = N_w.value, M_w.value, kappa_w.value
        omega, tipo = omega_w.value, grafo_w.value
        total = M ** N

        print(f"Exploración exhaustiva  —  Grafo={tipo}  N={N}  M={M}  κ={kappa}  ω*={omega}")
        print(f"Espacio: M^N = {M}^{N} = {total} configuraciones")

        if total > 20_000:
            print("⚠  Espacio demasiado grande. Reduce N o M.")
            return

        res = explorar(N, M, kappa, omega, tipo)
        conclusion_exhaustiva(res)

        if res["t_sync"]:
            t_s = res["t_sync"]
            fig, ax = plt.subplots(figsize=(7, 2.5))
            ax.hist(t_s, bins=range(min(t_s), max(t_s) + 2), align="left", rwidth=0.8)
            ax.set_xlabel("Tiempo de sincronización")
            ax.set_ylabel("Frecuencia")
            ax.set_title("Histograma de tiempos de sync")
            plt.tight_layout(); plt.show()


btn_sim.on_click(on_sim)
btn_exp.on_click(on_exp)


# ============================================================
# Panel
# ============================================================

panel = widgets.VBox([
    widgets.HTML("<h3>Simulador Discreto tipo Kuramoto sobre Grafos</h3>"
                 "<p style='font-size:12px'>"
                 "F<sub>κ</sub>(θ)<sub>i</sub> = θ<sub>i</sub> + ω* + σ<sub>κ</sub>(S<sub>i</sub>(θ)) mod M"
                 "</p>"),
    widgets.HBox([N_w, M_w]),
    widgets.HBox([kappa_w, omega_w]),
    widgets.HBox([T_w, grafo_w, theta0_w]),
    widgets.HBox([btn_sim, btn_exp]),
    salida,
])

display(panel)
```

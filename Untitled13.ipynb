```python
import pandas as pd
import requests
import plotly.express as px
 
from dash import Dash, dcc, html, dash_table
from dash.dependencies import Input, Output
 
import dash_bootstrap_components as dbc
 
# =========================
# API
# =========================
 
url = "https://www.economia.gob.mx/apidatamexico/tesseract/data.jsonrecords?Flow=1&Month=201501%2C201502%2C201503%2C201504%2C201505%2C201506%2C201507%2C201508%2C201509%2C201510%2C201511%2C201512%2C201601%2C201602%2C201603%2C201604%2C201605%2C201606%2C201607%2C201608%2C201609%2C201610%2C201611%2C201612%2C201701%2C201702%2C201703%2C201704%2C201705%2C201706%2C201707%2C201708%2C201709%2C201710%2C201711%2C201712%2C201801%2C201802%2C201803%2C201804%2C201805%2C201806%2C201807%2C201808%2C201809%2C201810%2C201811%2C201812%2C201901%2C201902%2C201903%2C201904%2C201905%2C201906%2C201907%2C201908%2C201909%2C201910%2C201911%2C201912%2C202001%2C202002%2C202003%2C202004%2C202005%2C202006%2C202007%2C202008%2C202009%2C202010%2C202011%2C202012%2C202101%2C202102%2C202103%2C202104%2C202105%2C202106%2C202107%2C202108%2C202109%2C202110%2C202111%2C202112%2C202201%2C202202%2C202203%2C202204%2C202205%2C202206%2C202207%2C202208%2C202209%2C202210%2C202211%2C202212%2C202301%2C202302%2C202303%2C202304%2C202305%2C202306%2C202307%2C202308%2C202309%2C202310%2C202311%2C202312%2C202401%2C202402%2C202403%2C202404%2C202405%2C202406%2C202407%2C202408%2C202409%2C202410%2C202411%2C202412%2C202501%2C202502%2C202503%2C202504%2C202505%2C202506%2C202507%2C202508%2C202509%2C202510%2C202511%2C202512&State=1%2C11%2C22%2C24&cube=economy_foreign_trade_ent&drilldowns=Chapter+4+Digit%2CCountry%2CFlow%2CMonth%2CState%2CYear&locale=es&measures=Trade+Value%22
 
response = requests.get(url)
 
if response.status_code != 200:
    raise Exception("Error al conectar con la API")
 
data_json = response.json()
 
df = pd.DataFrame(data_json["data"])
 
# =========================
# LIMPIEZA
# =========================
 
df["fecha"] = pd.to_datetime(
    df["Month ID"].astype(str) + "01",
    format="%Y%m%d",
    errors="coerce"
)
 
df["trade_value"] = pd.to_numeric(
    df["Trade Value"],
    errors="coerce"
)
 
df["estado"] = df["State"].astype(str)
df["pais"] = df["Country"].astype(str)
df["capitulo"] = df["Chapter 4 Digit"].astype(str)
 
df["anio"] = pd.to_numeric(
    df["Year"],
    errors="coerce"
)
 
df_trade = df[
    ["fecha", "anio", "estado", "pais", "capitulo", "trade_value"]
].dropna()
 
# =========================
# DASH APP
# =========================
 
app = Dash(
    __name__,
    external_stylesheets=[dbc.themes.BOOTSTRAP]
)
 
# =========================
# LAYOUT
# =========================
 
app.layout = dbc.Container([
 
    html.H1(
        "Dashboard Comercio Exterior Bajío",
        className="text-center my-4"
    ),
 
    dbc.Row([
 
        dbc.Col([
 
            html.Label("Rango de años"),
 
            dcc.RangeSlider(
                id="anio-slider",
                min=int(df_trade["anio"].min()),
                max=int(df_trade["anio"].max()),
                value=[
                    int(df_trade["anio"].min()),
                    int(df_trade["anio"].max())
                ],
                marks={
                    int(year): str(int(year))
                    for year in sorted(df_trade["anio"].unique())
                },
                step=1
            )
 
        ], width=8),
 
        dbc.Col([
 
            html.Label("Estado"),
 
            dcc.Dropdown(
                id="estado-dropdown",
                options=
                [{"label": "Todos", "value": "Todos"}] +
                [
                    {"label": e, "value": e}
                    for e in sorted(df_trade["estado"].unique())
                ],
                value="Todos",
                clearable=False
            )
 
        ], width=4)
 
    ]),
 
    html.Br(),
 
    dbc.Row([
 
        dbc.Col(
            dbc.Card(
                dbc.CardBody([
                    html.H4("Importaciones Totales"),
                    html.H2(id="total-importaciones")
                ]),
                color="primary",
                inverse=True
            ),
            width=4
        ),
 
        dbc.Col(
            dbc.Card(
                dbc.CardBody([
                    html.H4("Principal Proveedor"),
                    html.H2(id="pais-dominante")
                ]),
                color="success",
                inverse=True
            ),
            width=4
        ),
 
        dbc.Col(
            dbc.Card(
                dbc.CardBody([
                    html.H4("Índice HHI"),
                    html.H2(id="hhi-box")
                ]),
                color="warning",
                inverse=True
            ),
            width=4
        )
 
    ]),
 
    html.Br(),
 
    dbc.Row([
 
        dbc.Col(
            dcc.Graph(id="grafica1"),
            width=6
        ),
 
        dbc.Col(
            dcc.Graph(id="grafica2"),
            width=6
        )
 
    ]),
 
    html.Br(),
 
    dbc.Row([
 
        dbc.Col(
            dcc.Graph(id="grafica3"),
            width=6
        ),
 
        dbc.Col(
            dcc.Graph(id="grafica4"),
            width=6
        )
 
    ]),
 
    html.Br(),
 
    dash_table.DataTable(
        id="tabla",
        page_size=10,
        style_table={"overflowX": "auto"},
        style_cell={
            "textAlign": "left",
            "padding": "5px"
        }
    )
 
], fluid=True)
 
# =========================
# CALLBACK
# =========================
 
@app.callback(
 
    [
        Output("total-importaciones", "children"),
        Output("pais-dominante", "children"),
        Output("hhi-box", "children"),
        Output("grafica1", "figure"),
        Output("grafica2", "figure"),
        Output("grafica3", "figure"),
        Output("grafica4", "figure"),
        Output("tabla", "data"),
        Output("tabla", "columns")
    ],
 
    [
        Input("anio-slider", "value"),
        Input("estado-dropdown", "value")
    ]
 
)
 
def actualizar_dashboard(anio_range, estado):
 
    data = df_trade[
        (df_trade["anio"] >= anio_range[0]) &
        (df_trade["anio"] <= anio_range[1])
    ]
 
    if estado != "Todos":
        data = data[data["estado"] == estado]
 
    total_importaciones = "${:,.0f}".format(
        data["trade_value"].sum()
    )
 
    pais_top = (
        data.groupby("pais")["trade_value"]
        .sum()
        .sort_values(ascending=False)
        .index[0]
    )
 
    hhi_data = (
        data.groupby("pais")["trade_value"]
        .sum()
        .reset_index()
    )
 
    hhi_data["share"] = (
        hhi_data["trade_value"] /
        hhi_data["trade_value"].sum()
    )
 
    hhi = round(
        (hhi_data["share"] ** 2).sum(),
        3
    )
 
    # =========================
    # GRAFICA 1
    # =========================
 
    g1_data = (
        data.groupby("fecha")["trade_value"]
        .sum()
        .reset_index()
    )
 
    fig1 = px.line(
        g1_data,
        x="fecha",
        y="trade_value",
        title="Evolución Temporal"
    )
 
    # =========================
    # GRAFICA 2
    # =========================
 
    g2_data = (
        data.groupby("pais")["trade_value"]
        .sum()
        .reset_index()
        .sort_values(
            by="trade_value",
            ascending=False
        )
        .head(10)
    )
 
    fig2 = px.bar(
        g2_data,
        x="trade_value",
        y="pais",
        orientation="h",
        title="Top Países Proveedores"
    )
 
    # =========================
    # GRAFICA 3
    # =========================
 
    hhi_year = (
        data.groupby(
            ["anio", "pais"]
        )["trade_value"]
        .sum()
        .reset_index()
    )
 
    hhi_year["share"] = (
        hhi_year.groupby("anio")["trade_value"]
        .transform(lambda x: x / x.sum())
    )
 
    hhi_year = (
        hhi_year.groupby("anio")["share"]
        .apply(lambda x: (x ** 2).sum())
        .reset_index(name="HHI")
    )
 
    fig3 = px.line(
        hhi_year,
        x="anio",
        y="HHI",
        markers=True,
        title="Índice HHI"
    )
 
    # =========================
    # GRAFICA 4
    # =========================
 
    top_paises = (
        data.groupby("pais")["trade_value"]
        .sum()
        .sort_values(ascending=False)
        .head(10)
        .index
    )
 
    heatmap_data = (
        data[data["pais"].isin(top_paises)]
        .groupby(
            ["estado", "pais"]
        )["trade_value"]
        .sum()
        .reset_index()
    )
 
    fig4 = px.density_heatmap(
        heatmap_data,
        x="pais",
        y="estado",
        z="trade_value",
        title="Heatmap Estado - País"
    )
 
    tabla_data = data.to_dict("records")
 
    tabla_columns = [
        {"name": col, "id": col}
        for col in data.columns
    ]
 
    return (
        total_importaciones,
        pais_top,
        str(hhi),
        fig1,
        fig2,
        fig3,
        fig4,
        tabla_data,
        tabla_columns
    )
 
# =========================
# RUN
# =========================
 
if __name__ == "__main__":
    app.run(debug=True)
```

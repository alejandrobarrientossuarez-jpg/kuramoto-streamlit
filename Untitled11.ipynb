```python
import tkinter as tk
from tkinter import ttk, filedialog, messagebox
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg


class DashboardCostosEnergia:
    def __init__(self, root):
        self.root = root
        self.root.title("Dashboard Costos de Energía Industrial")
        self.root.geometry("1200x750")

        self.df_long = None
        self.df_filtrado = None
        self.canvas = None

        self.crear_interfaz()

    def crear_interfaz(self):
        titulo = tk.Label(
            self.root,
            text="Dashboard Comparativo de Costos de Energía Industrial",
            font=("Arial", 18, "bold")
        )
        titulo.pack(pady=10)

        subtitulo = tk.Label(
            self.root,
            text="Región Bajío | pesos x KH",
            font=("Arial", 11)
        )
        subtitulo.pack()

        btn_cargar = tk.Button(
            self.root,
            text="Cargar archivo Excel",
            command=self.cargar_excel,
            font=("Arial", 11),
            bg="#2E86C1",
            fg="white"
        )
        btn_cargar.pack(pady=10)

        self.frame_filtros = tk.Frame(self.root)
        self.frame_filtros.pack(fill="x", padx=10, pady=5)

        self.frame_metricas = tk.Frame(self.root)
        self.frame_metricas.pack(fill="x", padx=10, pady=5)

        self.frame_grafica = tk.Frame(self.root)
        self.frame_grafica.pack(fill="both", expand=True, padx=10, pady=10)

    def cargar_excel(self):
        ruta = filedialog.askopenfilename(
            title="Selecciona el archivo Excel",
            filetypes=[("Archivos Excel", "*.xlsx *.xls")]
        )

        if not ruta:
            return

        try:
            raw = pd.read_excel(ruta, header=None, engine="openpyxl")

            fila_anio = raw.iloc[2]
            fila_mes = raw.iloc[3]
            fila_indicador = raw.iloc[4]

            columnas = ["estado"]

            for i in range(1, len(raw.columns)):
                anio = str(fila_anio[i]).strip()
                mes = str(fila_mes[i]).strip()
                indicador = str(fila_indicador[i]).strip()
                columnas.append(f"{anio}_{mes}_{indicador}")

            df = raw.iloc[5:].copy()
            df.columns = columnas
            df.reset_index(drop=True, inplace=True)

            df["estado"] = df["estado"].astype(str)

            df_long = df.melt(
                id_vars="estado",
                var_name="variable",
                value_name="valor"
            )

            partes = df_long["variable"].str.split("_", n=2, expand=True)

            df_long["anio"] = partes[0]
            df_long["mes"] = partes[1]
            df_long["indicador"] = partes[2]

            df_long["valor"] = pd.to_numeric(df_long["valor"], errors="coerce")
            df_long = df_long.dropna(subset=["valor"])

            self.df_long = df_long

            self.crear_filtros()
            self.actualizar_dashboard()

        except Exception as e:
            messagebox.showerror("Error", f"No se pudo cargar el archivo:\n{e}")

    def crear_filtros(self):
        for widget in self.frame_filtros.winfo_children():
            widget.destroy()

        self.estado_var = tk.StringVar(value="Todos")
        self.anio_var = tk.StringVar(value="Todos")
        self.mes_var = tk.StringVar(value="Todos")
        self.indicador_var = tk.StringVar(value="Todos")
        self.grafica_var = tk.StringVar(value="Barras por estado")

        estados = ["Todos"] + sorted(self.df_long["estado"].unique())
        anios = ["Todos"] + sorted(self.df_long["anio"].unique())
        meses = ["Todos"] + sorted(self.df_long["mes"].unique())
        indicadores = ["Todos"] + sorted(self.df_long["indicador"].unique())

        self.crear_combo("Estado", self.estado_var, estados, 0)
        self.crear_combo("Año", self.anio_var, anios, 1)
        self.crear_combo("Mes", self.mes_var, meses, 2)
        self.crear_combo("Indicador", self.indicador_var, indicadores, 3)

        graficas = [
            "Barras por estado",
            "Evolución por año",
            "Comparación mensual",
            "Distribución por indicador"
        ]

        self.crear_combo("Gráfica", self.grafica_var, graficas, 4)

        btn_actualizar = tk.Button(
            self.frame_filtros,
            text="Actualizar",
            command=self.actualizar_dashboard,
            bg="#28B463",
            fg="white"
        )
        btn_actualizar.grid(row=1, column=5, padx=10, pady=5)

    def crear_combo(self, texto, variable, valores, columna):
        label = tk.Label(self.frame_filtros, text=texto)
        label.grid(row=0, column=columna, padx=8)

        combo = ttk.Combobox(
            self.frame_filtros,
            textvariable=variable,
            values=valores,
            state="readonly",
            width=20
        )
        combo.grid(row=1, column=columna, padx=8, pady=5)

    def filtrar_datos(self):
        df = self.df_long.copy()

        if self.estado_var.get() != "Todos":
            df = df[df["estado"] == self.estado_var.get()]

        if self.anio_var.get() != "Todos":
            df = df[df["anio"] == self.anio_var.get()]

        if self.mes_var.get() != "Todos":
            df = df[df["mes"] == self.mes_var.get()]

        if self.indicador_var.get() != "Todos":
            df = df[df["indicador"] == self.indicador_var.get()]

        self.df_filtrado = df

    def actualizar_dashboard(self):
        self.filtrar_datos()
        self.mostrar_metricas()
        self.mostrar_grafica()

    def mostrar_metricas(self):
        for widget in self.frame_metricas.winfo_children():
            widget.destroy()

        df = self.df_filtrado

        if df.empty:
            mensaje = tk.Label(
                self.frame_metricas,
                text="No hay datos con los filtros seleccionados.",
                font=("Arial", 12, "bold")
            )
            mensaje.pack()
            return

        promedio = round(df["valor"].mean(), 2)
        maximo = round(df["valor"].max(), 2)
        minimo = round(df["valor"].min(), 2)
        obs = len(df)

        metricas = [
            ("Promedio", promedio),
            ("Máximo", maximo),
            ("Mínimo", minimo),
            ("Observaciones", obs)
        ]

        for titulo, valor in metricas:
            frame = tk.Frame(
                self.frame_metricas,
                bd=2,
                relief="groove",
                padx=20,
                pady=10
            )
            frame.pack(side="left", padx=10, pady=5)

            tk.Label(
                frame,
                text=titulo,
                font=("Arial", 10, "bold")
            ).pack()

            tk.Label(
                frame,
                text=str(valor),
                font=("Arial", 14)
            ).pack()

    def mostrar_grafica(self):
        for widget in self.frame_grafica.winfo_children():
            widget.destroy()

        df = self.df_filtrado

        if df.empty:
            return

        fig, ax = plt.subplots(figsize=(10, 5))

        tipo = self.grafica_var.get()

        if tipo == "Barras por estado":
            tabla = df.groupby(["estado", "indicador"])["valor"].mean().unstack()
            tabla.plot(kind="bar", ax=ax)
            ax.set_title("Promedio de costos por estado")
            ax.set_xlabel("Estado")
            ax.set_ylabel("Valor")

        elif tipo == "Evolución por año":
            tabla = df.groupby(["anio", "estado"])["valor"].mean().unstack()
            tabla.plot(kind="line", marker="o", ax=ax)
            ax.set_title("Evolución promedio por año")
            ax.set_xlabel("Año")
            ax.set_ylabel("Valor")

        elif tipo == "Comparación mensual":
            tabla = df.groupby(["mes", "estado"])["valor"].mean().unstack()
            tabla.plot(kind="bar", ax=ax)
            ax.set_title("Comparación mensual promedio")
            ax.set_xlabel("Mes")
            ax.set_ylabel("Valor")

        elif tipo == "Distribución por indicador":
            df.boxplot(column="valor", by="indicador", ax=ax)
            ax.set_title("Distribución por indicador")
            ax.set_xlabel("Indicador")
            ax.set_ylabel("Valor")
            fig.suptitle("")

        ax.grid(True, alpha=0.3)
        plt.xticks(rotation=45)
        plt.tight_layout()

        self.canvas = FigureCanvasTkAgg(fig, master=self.frame_grafica)
        self.canvas.draw()
        self.canvas.get_tk_widget().pack(fill="both", expand=True)


if __name__ == "__main__":
    root = tk.Tk()
    app = DashboardCostosEnergia(root)
    root.mainloop()
```
